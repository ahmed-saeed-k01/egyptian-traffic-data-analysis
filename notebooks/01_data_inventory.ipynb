```
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(r"D:\Traffic_Data")
RAW_DIR = PROJECT_ROOT / "data" / "raw"
REPORT_DIR = PROJECT_ROOT / "reports" / "inventory"

REPORT_DIR.mkdir(parents=True, exist_ok=True)

raw_files = sorted(RAW_DIR.glob("*.xlsx"))

print(f"Excel files found: {len(raw_files)}")

for file in raw_files:
    print(file.name)
```

```
import pandas as pd

manifest = []

for file in raw_files:
    stat = file.stat()

    manifest.append({
        "file": file.name,
        "extension": file.suffix,
        "size_mb": round(stat.st_size / (1024 ** 2), 2),
        "modified_at": pd.Timestamp.fromtimestamp(stat.st_mtime)
    })

file_manifest = pd.DataFrame(manifest)

file_manifest
```

```
file_manifest.to_csv(
    REPORT_DIR / "file_manifest.csv",
    index=False
)
```
